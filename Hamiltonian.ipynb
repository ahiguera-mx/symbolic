### Lagrangian → Hamiltonian Discovery
* Based on Legendre Transform
$$
H(q, p) = \sum_i p_i \dot{q}_i - L(q, \dot{q})
$$

```python
import sympy as sp
```

```python
class HamiltonianDiscovery:
    def __init__(self, lagrangian, q, q_dot, t=sp.Symbol('t')):
        self.q = q
        self.q_dot = q_dot
        self.t = t
        self.L = lagrangian

        # Momentum definition: p = ∂L/∂q_dot
        self.p = sp.Matrix([sp.diff(self.L, qd) for qd in self.q_dot])

        # Try to solve for q_dot in terms of q and p (Legendre inversion)
        try:
            self.q_dot_solved = sp.solve(self.p - sp.MatrixSymbol('p', len(self.q), 1), self.q_dot, dict=True)[0]
        except IndexError:
            raise ValueError("Could not invert for q_dot. Lagrangian may be singular.")

        # Substitute q_dot in L to get H = p*q_dot - L
        p_syms = sp.Matrix(sp.symbols(f'p:{len(self.q)}'))
        q_dot_subs = [self.q_dot_solved[qd] for qd in self.q_dot]
        H_expr = sum(p_syms[i] * q_dot_subs[i] for i in range(len(q))) - self.L.subs(self.q_dot_solved)

        self.H = sp.simplify(H_expr)

    def get_hamiltonian(self):
        return self.H

    def __repr__(self):
        return f"Hamiltonian: H(q, p) = {self.H}"
```

```python
# Symbols
t = sp.Symbol('t')
q = [sp.Function('x')(t)]
q_dot = [q[0].diff(t)]
```

* Example harmonic oscillator 

```python
# Mass and spring constants
m, k = sp.symbols('m k')
```

```python
# Lagrangian for harmonic oscillator
L = (1/2)*m*q_dot[0]**2 - (1/2)*k*q[0]**2
```

```python
# Run conversion
converter = HamiltonianDiscovery(L, q, q_dot, t)
H = converter.get_hamiltonian()
```
